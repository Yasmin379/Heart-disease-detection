### **Feature Distribution**

```python
features = df.columns.drop("heart_disease")

for column in features:
    plt.figure(figsize=(8, 4))
    sns.histplot(data=df, x=column, bins=20, kde=True)
    plt.title(f"Distribution of {column}")
    plt.show()
```

```python
df.skew(numeric_only=True)
```

### **Outlier Analysis**
---

```python
for column in features:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[column])
    plt.title(f"Box Plot of {column}")
    plt.show()
```

```python
outlier_summary = []

for column in features:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df[column] < lower) | (df[column] > upper)).sum()
    outlier_summary.append([column, lower, upper, count])

pd.DataFrame(outlier_summary, columns=["Feature", "Lower Bound", "Upper Bound", "Outlier Count"])
```

```python
import sys
sys.path.append("..")
from src.preprocessing import preprocess_data

X_train, X_test, y_train, y_test = preprocess_data()

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```
